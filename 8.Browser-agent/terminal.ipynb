# 1. Simple command

```python
import subprocess

command = "pwd"

result = subprocess.run(
    command,
    shell=True,          # Run the command through the system shell
    capture_output=True, # Capture stdout and stderr
    text=True,           # Return output as strings instead of bytes
    timeout=30,          # Stop the command if it runs for more than 30 seconds
)
```

```python
print(result.stdout)
```

```python
print(result.stderr)
```

```python
print(result.returncode)
```

```python
print(result)
```

# 2. Command that produces an error

```python
command = "ls /random-file"

result = subprocess.run(
    command,
    shell=True,
    capture_output=True,
    text=True,
    timeout=30,
)

print("STDOUT:", result.stdout)
print("STDERR:", result.stderr)
print("RETURN CODE:", result.returncode)
```

# 3. Try Python

```python
command = "python --version"

result = subprocess.run(
    command,
    shell=True,
    capture_output=True,
    text=True,
    timeout=30,
)

print(result.stdout)
print(result.stderr)
```

# 4. Try a pipe

```python
command = "ls -la | grep .py"

result = subprocess.run(
    command,
    shell=True,
    capture_output=True,
    text=True,
    timeout=30,
)

print(result.stdout)
```

# 5. Try a timeout

```python
command = "sleep 5"

result = subprocess.run(
    command,
    shell=True,
    capture_output=True,
    text=True,
    timeout=3,
)
```
