```python
from langchain_google_community import CalendarToolkit

from langchain_google_community.calendar.utils import (
    build_calendar_service,
    get_google_credentials,
)
```

```python
credentials = get_google_credentials(
    token_file="token.json",
    scopes=["https://www.googleapis.com/auth/calendar"],
    client_secrets_file="credentials.json",
)
```

```python
calendar_resource = build_calendar_service(credentials=credentials)
calendar_toolkit = CalendarToolkit(api_resource=calendar_resource)
```

```python
tools = calendar_toolkit.get_tools()

for t in tools:
    print(t.name)
```

```python
for tool in tools:
    print(f"\nTool: {tool.name}")

    if tool.args_schema:
        for name, field in tool.args_schema.model_fields.items():
            print(f"  {name}: {field.annotation}")
    else:
        print("  No arguments")
```

```python
tool_map = {t.name: t for t in tools}

tool_map['create_calendar_event'].invoke(
    {
        "summary": "Calculus exam",
        "start_datetime": "2026-07-4 11:00:00",
        "end_datetime": "2026-07-4 13:00:00",
        "timezone": "America/Mexico_City",
        "location": "UAM Cuajimalpa",
        "description": "Event created from the LangChain toolkit",
        "reminders": [{"method": "popup", "minutes": 60}],
        "conference_data": True,
        "color_id": "5",
    }
)
```

# Gmail Toolkit

```python
from langchain_google_community import CalendarToolkit, GmailToolkit

from langchain_google_community.calendar.utils import (
    build_calendar_service,
    get_google_credentials,
)

from langchain_google_community.gmail.utils import (
    build_gmail_service
)
```

```python
credentials = get_google_credentials(
    token_file="token.json",
    scopes=["https://www.googleapis.com/auth/calendar",
            "https://mail.google.com/",
           ],
    client_secrets_file="credentials.json",
)
```

```python
gmail_resource = build_gmail_service(credentials=credentials)
gmail_toolkit = GmailToolkit(api_resource=gmail_resource)
tools = gmail_toolkit.get_tools()
```

```python
for t in tools:
    print(t.name)
```
