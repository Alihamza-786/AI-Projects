```python
import os, json, csv, time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from pathlib import Path
from rich import print as rprint
from rich.table import Table
from rich.panel import Panel

# ── Set your data directory ───────────────────────────────────
DATA_DIR = "./space_data"   # adjust if needed
# ─────────────────────────────────────────────────────────────

# Verify files exist
expected_files = [
"Space_Exploration_RAG_Document.docx"
]

print("Checking sample data files:")
for f in expected_files:
    path = Path(DATA_DIR) / f
    status = "✅" if path.exists() else "❌ MISSING"
    size   = f"{path.stat().st_size / 1024:.1f} KB" if path.exists() else ""
    print(f"  {status}  {f:40s}  {size}")
```

```python
# ─────────────────────────────────────────────────────────────
# 2.3  DOCX Loader — extracts text from Word documents
# ─────────────────────────────────────────────────────────────
from langchain_community.document_loaders import Docx2txtLoader

docx_loader = Docx2txtLoader(f"{DATA_DIR}/Space_Exploration_RAG_Document.docx")
docx_docs   = docx_loader.load()

print("=== DOCX Loader ===")
print(f"Documents loaded : {len(docx_docs)}")
print(f"Characters       : {len(docx_docs[0].page_content):,}")
print(f"Metadata         : {docx_docs[0].metadata}")
print(f"\nPreview:\n{docx_docs[0].page_content[:400]}")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter


full_text = docx_docs[0].page_content
metadata = docx_docs[0].metadata

print("Total Characters", len(full_text))

recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    separators=["\n\n", "\n", ". ", " ", ""],
    length_function=len,
)


# Create chunks

recursive_chunks = recursive_splitter.create_documents(
    texts=[full_text],
    metadatas=[metadata]
)

#  Debug output
print("=== Recursive Character Splitter ===")
print(f"Total chunks: {len(recursive_chunks)}")

sizes = [len(c.page_content) for c in recursive_chunks]
print(f"Avg size    : {sum(sizes)//len(sizes)} chars")
print(f"Min / Max   : {min(sizes)} / {max(sizes)} chars")

print("\nChunk 3 preview:\n")
print(recursive_chunks[2].page_content)
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
embeddings = OllamaEmbeddings(
    model="mxbai-embed-large:latest"   
)
```

```python
vector = embeddings.embed_query("What is space exploration?")
print(len(vector))
```

```python
texts = [chunk.page_content for chunk in recursive_chunks]
metadatas = [chunk.metadata for chunk in recursive_chunks]
```

```python
print(len(texts))
print(len(metadatas))
```

```python
print(metadatas[0])
```

```python
vectorstore = FAISS.from_texts(
    texts=texts,
    embedding=embeddings,
    metadatas=metadatas
)
```

## 1. Question:

When did the Space Age begin and what event marked it?

Expected Answer:
The Space Age began on October 4, 1957, when the Soviet Union launched Sputnik 1, the first artificial satellite, into Earth’s orbit.

```python
query = "When did the Space Age begin and what event marked it?"

results = vectorstore.similarity_search(query, k=2)

for i, doc in enumerate(results):
    print(f"\n--- Chunnk {i+1} ---")
    print(doc.page_content[:300])
```

# LLM

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="qwen2.5",
    temperature=0.2
)
```

```python
query = "When did the Space Age begin and what event marked it?"

# Step 1: retrieve context
docs = vectorstore.similarity_search(query, k=3)

context = "\n\n".join([d.page_content for d in docs])

# Step 2: prompt Qwen2.5
prompt = f"""
You are a helpful assistant. Answer using only the context below.

Context:
{context}

Question:
{query}

Answer:
"""

response = llm.invoke(prompt)

print(response.content)
```

```python
texts2 = [chunk.page_content for chunk in recursive_chunks]
metadatas2 = [chunk.metadata for chunk in recursive_chunks]
print(len(texts2))
print(len(metadatas2))
```

```python
print(metadatas2[0])
```

```python
vectorstore.add_texts(
    texts=texts2,
    metadatas=metadatas2
)
```

```python
faiss_index = vectorstore.index

vectors = faiss_index.reconstruct_n(
    0,
    faiss_index.ntotal
)

print(vectors.shape)
```

# PCA

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
points = pca.fit_transform(vectors)

print(points.shape)
```

```python
docstore = vectorstore.docstore._dict
index_to_docstore_id = vectorstore.index_to_docstore_id

labels = [
    docstore[index_to_docstore_id[i]].metadata.get("source", "unknown")
    for i in range(vectorstore.index.ntotal)
]
```

```python
import numpy as np
import matplotlib.pyplot as plt

unique_labels = list(set(labels))
color_map = {label: i for i, label in enumerate(unique_labels)}

colors = np.array([color_map[l] for l in labels])

plt.figure(figsize=(9, 7))

plt.scatter(
    points[:, 0],
    points[:, 1],
    c=colors,
    cmap="tab10",
    alpha=0.8
)

plt.title("FAISS 2-Document Embedding Visualization")
plt.show()
```
